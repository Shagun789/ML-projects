# Machine learning

Our target variable is fatalities and our preedictors are aboard, ground, Time_hour, Year, Month

# We will only takr numeric features


```python
features = ["Aboard","Ground","Time_hour"] #features
x = df[features]
y = df["Fatalities"] #target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=42)
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## We will use Random Forest Regressor as it is really uselful in handling non liner relationships
## less overfitting as compared to decision trees
## Handles feature interactions well

```python
#Train the model
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(
    n_estimators = 100,
    random_state = 42)
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
#Evaluate the model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("MAE-", mae)
print("MSE", mse)
print("R2 Score", r2)
```

```python
importance = pd.Series(model.feature_importances_, index=["Aboard","Ground","Time_hour"])
importance.sort_values().plot(kind = "barh")
import pandas as pd
```

```python
#it shows that passsengers onboard is the biggest factor affecting fatalities
```

```python
new = [[150,3,14]]
new = scaler.transform(new)
prediction = model.predict(new)
print("Predicted Fatalities-", prediction)
```
